## Data Attributes

* Make - Company of the vehicle

* Model - Car Model
    * 4WD/4X4 = Four-wheel drive 
    * AWD = All-wheel drive 
    * FFV = Flexible-fuel vehicle 
    * SWB = Short wheelbase 
    * LWB = Long wheelbase 
    * EWB = Extended wheelbase


* Vehicle Class - Class of vehicle depending on their utility, capacity and weight

* Engine Size(L) - Size of engine used in Litre

* Cylinders - Number of cylinders

* Transmission - Transmission type with number of gears
    * A = Automatic
    * AM = Automated manual
    * AS = Automatic with select shift
    * AV = Continuously variable
    * M = Manual
    * 3 to 10 = Number of gears

* Fuel Type: X = Regular gasoline(petrol), Z = Premium gasoline, D = Diesel, E = Ethanol(E85), N = Natural gas(CNG/LNG)

* Fuel Consumption City (L/100 km) - Fuel consumption in city roads (L/100 km)

* Fuel Consumption Hwy (L/100 km) - Fuel consumption in highways (L/100 km)

* Fuel Consumption Comb (L/100 km) - The combined fuel consumption (55% city, 45% highway) is shown in L/100 km

* Fuel Consumption Comb (mpg) - The combined fuel consumption in both city and highway is shown in mile per gallon(mpg)

* City and highway fuel consumption ratings are shown in litres per 100 kilometres (L/100 km) - the combined rating (55% city, 45% hwy) is shown in L/100 km and in miles per gallon (mpg)

* CO2 Emissions(g/km) - The tailpipe emissions of carbon dioxide (in grams per kilometre) for combined city and highway driving

Let's take a quick look at the DataFrame:


Let's check the dataset structure;\
checking column types, non-null counts and memory usage, to spot missing values and columns stored as the wrong dtype before any cleaning:

```python
co2_df.info()
```

**Observation:**
- We have 12 columns.
- There are five **`str`** columns: `Make`, `Model`, `Vehicle Class`, `Transmission`, and `Fuel Type`
- There are four **`float64`** columns: `Engine Size(L)`, `Fuel Consumption City (L/100 km)`, `Fuel Consumption Hwy (L/100 km)`, and\
`Fuel Consumption Comb (L/100 km)`
- There are three **`int64`** columns: `Cylinders`, `Fuel Consumption Comb (mpg)`, and `CO2 Emissions(g/km)`
- There are no missing values values in the dataset.

**Next:** Drop unnecessary columns:

We have two same columns `Fuel Consumption Comb (mpg)` and `Fuel Consumption Comb (L/100 km)` in different units.\
Since other columns are in **`L`**, let's drop `Fuel Consumption Comb (mpg)`:

```python
co2_df = co2_df.drop('Fuel Consumption Comb (mpg)', axis=1)
```

Now we have 11 columns including `Fuel Consumption Comb` column following the consistent unit `litres per 100 kilometres`.

**Next:** Let's check if we have duplicates:

```python
co2_df.duplicated().sum()
```

We have `1103` duplicate rows whcih is `~15%` of the dataset. I can’t tell whether they’re actual errors or just legitimate repeated records since
there is no additional information. So, it's resonable to drop them because in train-test split the same record could end up in both sets which will make our test score look better than it really is, because the model has already seen those exact rows.

**Next:** Let's drop duplicate rows:

```python
co2_df = co2_df.drop_duplicates()
```

```python
# Verify duplicates are gone
print("duplicates:",co2_df.duplicated().sum())
print("co2_df.shape:",co2_df.shape)
```

After dropping the duplicates there are 7,385 → 6,282 rows left. Which means if some duplicates were real separate cars, we removed real observations.
